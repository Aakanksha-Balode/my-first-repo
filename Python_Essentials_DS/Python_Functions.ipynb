### Defining and Calling Functions

```python
# Function Definition
def greet():
    print("Hello, there!")
```

```python
# calling a function
greet()
```

```python
# Function Parameters
def greet(name):
    print("Hello, " + name + "!")
```

```python
# Calling a Function with Arguments
greet("Alice")
```

```python
# Function Return Value
def multiply(a,b):
    product = a * b
    return product
result = multiply(5,3)
print(result)
```

```python
# Schema Example
def calculate_area(length, width):
    area = length * width
    return area
length = 5
width = 3
result = calculate_area(length, width)
print(result)
```

### Function Arguments and Return Values

```python
# Function Arguments
# Positional Arguments
def greet(name, age):
    print("Hello, " + name + "! You are " + str(age) + " years old.")
greet("Alice", 25)
```

```python
# Keyword Arguments
def greet(name, age):
    print("Hello, " + name + "! You are " + str(age) + " years old.")
greet(age = 30, name = "Bob")
```

```python
# Default Arguments
def greet(name, greeting = "Hello"):
    print(greeting + ", " + name + "!")
greet("Alice")
greet("Bob", greeting = "Hi")
```

```python
# Return Values
def multiply(a,b):
    product = a * b
    return product
result = multiply(5,3)
print(result)
```

```python
def calculate(a,b):
    sum = a + b
    difference = a - b
    return sum, difference
result_sum, result_difference = calculate(10,5)
print(result_sum)
print(result_difference)
```

```python
# Schema Example
def calculated_rectangle_properties(length, width):
    area = length * width
    perimeter = 2 * (length + width)
    return area, perimeter
```

#### Lambdas: Anonymous Functions

```python
## Syntax of Lambdas
# The syntax for defining a lambda function in Python is as follows:
lambda arguments: expression
```

```python
square = lambda x:x **2 
```

```python
## Using Lambdas
# Sorting: Lambdas can be used as key functions in sorting operations to define custom sorting criteria.
fruits = ['apple', 'banana', 'cherry', 'date']
sorted_fruits = sorted (fruits, key=lambda x: len(x))
```

```python
# Filtering: Lambdas can be used in combination with built-in functions like 'filter()' to filter elements from a list based on specific conditions.
numbers = [1,2,3,4,5,6]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
```

```python
# Mapping: Lambdas can be used with functions like 'map()' to apply a specific operation to each element of a list.
numbers = [1,2,3,4,5]
doubled_numbers = list(map(lambda x: x * 2, numbers))
```

### Understanding Scope of Variables in Functios

```python
## Local Scope
def my_function():
    x = 10
    print(x)
my_function() # output: 10
print(x)      #Error: NameError: x is not defined
```

```python
## Global Scope
x = 10
def my_function():
    print(x)
my_function()
print(x)
```

```python
## Modifying Global Variables within a Function
x = 10
def modify_global():
    global x
    x = 20
modify_global()
print(x)
```

```python
## Schema Example
total = 0
def calculate_bonus(sales):
    global total
    bonus = sales * 0.1
    total += sales
employee1_sales = 5000
employee1_bonus = calculate_bonus(employee1_sales)
print("Employee 1 Bonus:", employee1_bonus)
print("Total Sales:", total)
```

Conclusion:-
Understanding the scope of variables within functions is essential for writing robust and bug-free Python code.
Local variables are accessible only within the function where they are defined, while global variables can be accessed from anywhere in the program.
Modifying global variables within functions requires explicity declaring them as global.
By grasping the concept of variable scope, you can write code that is easier to understand, maintain, and debug.
