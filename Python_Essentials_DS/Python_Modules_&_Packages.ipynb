#### How to Import a Module?
#### The import Statement

```python
# Importing the math module
import math
# Using math module functions
result = math.sqrt(25)
print(result)
```

#### The from...import Statement

The 'from...import' statement allows you to import specific objects or functions from a module, eliminating the need to use the module name as a prefix. This provides a more concise and direct way to accessing module functionality.

```python
# Importing the sqrt function directly from the math module
from math import sqrt
# Using the sqrt function directly
result = sqrt(25)
print(result) 
```

 Multiple objects or functions can be imported using the 'from...import' statement by separating them with commas.

```python
# Importing multiple functions from the math module
from math import sqrt, sin, cos
relsut1 = sqrt(25)
result2 = sin(0.5)
result3 = cos(1.2)
```

#### The import...as Statement

```python
# Importing the math module and assigning it an alias 'm'
import math as m
# Using the sqrt function using the module alias
result = m.sqrt(25)
print(result)
```

### Some Essential Python Standard Libraries

#### 'math' Library
The 'math' library provides mathematical functions and constants for performing advanced mathematical operations. It includes functions for basic arithmetic, trigonometry, logarithmic operations and more.

```python
import math

# calculate the square root of number
result = math.sqrt(25)
print(result)

# calculate the value of pi
pi = math.pi
print(pi)
```

#### 'datetime' Library
The 'datetime' library provides classes for working with dates, times, and time intervals. It allows you to perform operations such as date and time calculations, formatting and parsing.

```python
import datetime

# Get the current date and time
current_datetime = datetime.datetime.now()
print(current_datetime)

# Format a datetime object
formatted_datetime = current_datetime.strftime("%Y-%m-%d %H:%M:%S")
print(formatted_datetime)
```

#### 'random' Library
The 'random' library provides functions for generating random numbers, selecting random elements and shuffling sequences. It is commonly used for tasks like geenerating random passwords, simulating games and data shuffling.

```python
import random

# Generate a random number between 1 and 10
random_number = random.randint(1,10)
print(random_number)

# Select a random element from a list
my_list = ['','','','']
random_element = random.choice(my_list)
print(random_element)

# Shuffle a list
random.shuffle(my_list)
print(my_list)
```

#### 'os' Library
The 'os' library provides a way to interact with the underlying operating system. It allows you to perform operations such as file and directory manipulation, environment variable access and process management.

```python
import os

# Get the current working directory
current_dir = os.getcwd()
print(current_dir)

# Create a new directory
new_dir = os.path.join(current_dir, 'new_directory')
os.mkdir(new_dir)

# Check if a file exists
file_path = os.path.join(current_dir, 'myfile.txt')
if os.path.exists(file_path):
    print("file exists")
else:
    print("file does not exists")
```

### Creating Your First Python Module

#### Creating Your First Python Module
#### Module Structure

To create a Python module, follow these steps:

- step 1: Create a new file with a '.py' extension. For example, 'my_module.py'.
- step 2: Open the file in a text editor or integrated development environment(IDE).
- step 3: Define the code you want to include in the module. This can be function definitions, variable assignments or class definitions.

```python
# my_module.py
def greet(name):
    print("Hello, " + name + "!")
def add_numbers(a,b):
    return a + b
```

#### Using Your Module
To use the functions defined in your module, you need to import the module in another Python program.

```python
import my_module
my_module.greet("Alice")
result = my_module.add_numbers(3,5)
print(result)
```

### Package Creation and Use
#### Package Structure
A package is a directory that contains multiple Python modules.
#### Package Creation
To create a package, you simply need to create a directory and include the '__init__.py' file within it. Here's an example:

my_package/

__init__.py

module1.py

module2.py
#### Using Modules from a Package
To use modules from a package, you can import them using the appropriate import statements.

```python
from my_package import module1
from my_package.subpackage1 import module3
module1.function1()
module3.function2()
```

#### Importing a Package

```python
import my_package
my_package.module1.function1()
my_package.subpackage1.module3.function2()
```

#### Importing Specific Objects from Modules

```python
from my_package.module import function1
from my_package.subpackage1.module3 import function2
function1()
function2()
```
