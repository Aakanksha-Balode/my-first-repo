```python
# Code Blocks and Nested Blocks
def greet(name):
    # Outer code block
    print("Hello, " + name + "!")
    if name == "Alice":
        # Nested code block
        print("You are amazing, " + name + "!")
    else:
        # Another nested code block
        print("Nice to meet you, " + name + "!")
    # Code outside the if-else block
    print("Have a greet day, " + name + "!")
# Function call
greet("Bob")
```

```python
### Basic Python Data Types and Their Operations
## Numbers

# Integer
x = 10 
y = 5
print(x + y)

# Floating-point number
a = 3.14
b = 2.5
print(a  * b)

# Complex number
c = 2 + 3j
d = 1 - 2j
print(c * d)
```

```python
## String
# Concatenation
first_name = "John"
last_name = "Doe"
full_name = first_name + " " + last_name
print(full_name)

# String slicing
message = "Hello, World!"
print(message[0:5])

# String formatting
age = 25
print("I am {} years old." .format(age))
```

```python
## Lists

# Creating a list
numbers = [1,2,3,4,5]

# Accessing elements
print(numbers[0])

# Modifying elements
numbers[0] = 10
print(numbers)

# Adding elements
numbers.append(6)
print(numbers)

# Removing elements
numbers.remove(3)
print(numbers)
```

```python
## Tuples

# Tuples are similar to lists, but they are immutable, meaning their elements cannot be modified after creation.
# Tuples are defined using parentheses (()).

# Creating a tuple
coordinates = (3,4)

# Accessing elements
x = coordinates[0]
y = coordinates[1]

print("x =",x)
print("y =",y)
```

```python
## Sets

# Creating a set
fruits = {"apple", "banana","orange"}

# Adding elements
fruits.add("grape")
print(fruits)

# Removing elements
fruits.remove("banana")
print(fruits)
```

```python
## Dictionaries

# Creating a dictionary
person = {
    "name": "",
    "age": 25,
    "city": "London"
}

# Accessing values
print(person["name"])
print(person.get("age"))

# Modifying values
person["age"] = 30
print(person)
```
