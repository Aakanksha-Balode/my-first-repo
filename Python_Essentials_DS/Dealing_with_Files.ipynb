### Reading from and Writing to Files
#### Reading from Files
file = open( ' data.txt ' ,  ' r ' )
- read(): Reads the entire contents of the file as a single string.
- readline(): Reads a single line from the file and returns it as a string.
- readlines(): Reads all lines from the file and returns them as a list of strings.

```python
file = open('data.txt', 'r')
lines = file.readlines()
file.close()
for line in lines:
    print(line)
```

#### Writing to Files

```python
file = open('output.txt', 'w')
```

Once the file is opened n write mode, you can use the 'write()' method to write data to the file.

```python
file = open('output.txt', 'w')
file.write('Hello, world!\n')
file.write('This is a new line.')
file.close()
```

#### Context Managers and 'with' Statement

```python
with open('data.txt', 'r') as file:
    lines = file.readlines()
for line in lines:
    print(line)
```

### File Handling for CSV and JSON Files
#### CSV File Handling

```python
import csv 
with open('data.csv', 'r') as file:
    reader = csv.reader(file)
    for row inn reader:
    print(row)
```

To write data to a csv file, you can use the 'csv.writer' object.

```python
import csv
data =[
    ['name', 'age', 'country'],
    ['alice', 25, 'uk'],
    ['bob', 37, 'usa']
]
with open('output.csv', 'w', newline= '') as file:
    writer = csv.writer(file)
    writer.writerows(data)
```

CSV files can also have headers, specifying the column names. The 'csv.DictReader' and 
'csv.DictWriter' objects can be used to read and write CSV files with headers, respectively.
These objects allow you to access data using column names instead of indices.

#### JSON File Handling
JSON files are commonly used for storing structured data, following a key-value pair format similar to Python dictionaries. Python provides the 'json' module, making it easy to work with JSON files.

```python
import json
with open('data.json', 'r') as file:
    data = json.load(file)
    print(data)
```

To write data to a JSON file, you can use the 'json.dump()' function. 

```python
import json
data = {
    'name': 'Alice',
    'age': 25,
    'country': 'UK'
}
with open('output.json', 'w') as file:
    json.dump(data, file)
```

#### Schema Example

```python
import csv
import json
employees = []
with open('employees.csv', 'r') as csv_file:
    reader = csv.DictReader(csv_file)
    for row in reader:
        employees.append(row)
with open('employees.json', 'w') as json_file:
    json.dump(employees, json_file)
```

### Handling Exceptions during File Operations
#### Understanding Exceptions in File Handling
During file operations, various exceptions can occur, such as:
- FileNotFoundError : Raised when a file is not found or cannot be accessed.
- PermissionError :
- IOError :
- ValueError :

#### Using try-except Blocks for Exception Handling

```python
try:
    file = open('data.txt', 'r')
    # Perform file operations here
except FileNotFoundError:
    print("File not Found.")
except PermissionError:
    print("permission denied")
except IOError:
    print("An I/O error occurred")
except Exception as e:
    print("An unexpected error occurred:", str(e))
finally:
    file.close()
```

```python
try:
    with open('data.txt', 'r') as file:
        for line in file:
            try:
                values = line.split('.')
                num1 = int(values[0])
                num2 = int(values[1])
                result = num1/num2
                print("Result:", result)
            except ValueError:
                print("Invalid value encontered")
            except ZeroDivisionError:
                print("Cannot divide by zero")
            except Exception as e:
                print("An unexpected error occurred:", str(e))
except FileNotFoundError:
    print("File not found")
except PermissionError:
    print("Permission denied")
except IOError:
    print("An I/O error occurred")
except Exception as e:
    print("An unexpected error occurred:", str(e))
```
